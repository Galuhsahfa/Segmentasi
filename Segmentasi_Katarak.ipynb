```python
# Install OpenCV (jika belum terinstall)
!pip install opencv-python

# Import pustaka yang dibutuhkan
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Fungsi untuk menampilkan gambar
def show_image(title, img, cmap='gray'):
    plt.imshow(img, cmap=cmap)
    plt.title(title)
    plt.axis('off')
    plt.show()

# Load gambar (gunakan gambar Anda sendiri di Colab)
# Ganti path ini dengan path ke gambar Anda
image_path = 'eye_cataract_sample.jpg'  # Upload gambar ke Colab
img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

# Cropping (dari 1280x1280 ke 1000x1000)
cropped_img = img[140:1140, 140:1140]
show_image("Cropped Image", cropped_img)

# Thresholding untuk membuat citra biner
_, binary_img = cv2.threshold(cropped_img, 100, 255, cv2.THRESH_BINARY)
show_image("Thresholding", binary_img)

# Structuring element
kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5, 5))

# Morphological Closing (dilasi lalu erosi)
closing = cv2.morphologyEx(binary_img, cv2.MORPH_CLOSE, kernel)
show_image("After Closing", closing)

# Morphological Opening (erosi lalu dilasi)
opening = cv2.morphologyEx(closing, cv2.MORPH_OPEN, kernel)
show_image("After Opening", opening)

# Simpan hasil akhir
cv2.imwrite("hasil_segmentasi_katarak.jpg", opening)
```
